```python
import httpx
```

```python
BASE_URL = "http://127.0.0.1:8000"

health_resp = httpx.get(f"{BASE_URL}/health")
health_resp.json()
```

### 注册

```python
auth_register_resp = httpx.post(f"{BASE_URL}/api/auth/register", json={
    "username": "limingxuan",
    "password": "Li314159_",
    "nickname": "黎明轩"
})
auth_register_resp.json()
```

### 登录

```python
login_resp = httpx.post(
    f"{BASE_URL}/api/auth/login",
    json={"username": "limingxuan", "password": "Li314159_"},
)
login_data = login_resp.json()
login_data
```

```python
session = httpx.Client(base_url=BASE_URL)
session.headers.update({"Authorization": f"Bearer {login_data["data"]["access_token"]}"})
# session.headers.update({"Authorization": f"Bearer fake_token"})
```

### 获取个人信息

```python
uesr_me_resp = session.get("/api/user/me")
uesr_me_resp.json()
```
