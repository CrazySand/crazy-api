```python
import requests
```

```python
BASE_URL = "http://127.0.0.1:8000"

resp = requests.get(f"{BASE_URL}/health")
resp.json()
```

### 注册

```python
resp = requests.post(f"{BASE_URL}/api/auth/register", json={
    "username": "limingxuan2",
    "password": "Li314159_",
    "nickname": "1"
})
resp.json()
```

### 登录

```python
login_resp = requests.post(
    f"{BASE_URL}/api/auth/login",
    json={"username": "limingxuan2", "password": "Li314159_"},
)
login_data = login_resp.json()
login_data
```

```python
session = requests.Session()
session.headers.update({"Authorization": f"Bearer {login_data["data"]["access_token"]}"})
# session.headers.update({"Authorization": f"Bearer fake_token"})
```

### 获取个人信息

```python
resp = session.get(f"{BASE_URL}/api/user/me")
resp.json()
```
